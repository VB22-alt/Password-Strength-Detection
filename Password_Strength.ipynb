```python
# password_strength_predictor.py

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# ----------------------------
# STEP 1: Sample password dataset
# ----------------------------
data = {
    'password': [
        '12345', 'password', 'letmein', 'Hello123', 'Riddhi@2003',
        'abc', 'Admin@123', 'qwerty', 'P@ssw0rd', 'mydogname',
        'ILoveYou!', '654321', 'Test@1234', 'sunshine', 'Happy#2025'
    ],
    'strength': [
        'Weak', 'Weak', 'Weak', 'Medium', 'Strong',
        'Weak', 'Medium', 'Weak', 'Strong', 'Weak',
        'Medium', 'Weak', 'Strong', 'Weak', 'Strong'
    ]
}

df = pd.DataFrame(data)

# ----------------------------
# STEP 2: Feature extraction
# ----------------------------
def extract_features(password):
    length = len(password)
    digits = sum(c.isdigit() for c in password)
    upper = sum(c.isupper() for c in password)
    lower = sum(c.islower() for c in password)
    special = sum(not c.isalnum() for c in password)
    return [length, digits, upper, lower, special]

features = [extract_features(p) for p in df['password']]
X = pd.DataFrame(features, columns=['length', 'digits', 'upper', 'lower', 'special'])

# Encode labels
y = df['strength'].map({'Weak': 0, 'Medium': 1, 'Strong': 2})

# ----------------------------
# STEP 3: Train/test split
# ----------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# ----------------------------
# STEP 4: Train model
# ----------------------------
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# ----------------------------
# STEP 5: Evaluate
# ----------------------------
y_pred = model.predict(X_test)
print("Model accuracy:", accuracy_score(y_test, y_pred))

# ----------------------------
# STEP 6: Predict user input
# ----------------------------
def predict_strength(password):
    features = extract_features(password)
    prediction = model.predict([features])[0]
    labels = {0: 'Weak', 1: 'Medium', 2: 'Strong'}
    return labels[prediction]

# ----------------------------
# STEP 7: Test interactively
# ----------------------------
while True:
    pwd = input("\nEnter a password to check its strength (or type 'exit'): ")
    if pwd.lower() == 'exit':
        break
    print("Predicted strength:", predict_strength(pwd))
```
